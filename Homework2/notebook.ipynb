```python
import matplotlib.pyplot as plt
df['fuel_efficiency_mpg'].hist(bins=40,edgecolor='black')
plt.xlabel('Fuel Efficiency (mpg)')
plt.ylabel('Count')
plt.title('Distribution of Fuel Efficiency')
plt.show()
skew=df['fuel_efficiency_mpg'].skew()
print(f'Skewness of fuel_efficiency_mpg: {skew:.2f}')
```
